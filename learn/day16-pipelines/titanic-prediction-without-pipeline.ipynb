```python
import pandas as pd
import numpy as np
import pickle
```

```python
ohe_sex = pickle.load(open('/kaggle/input/models/sanjanbm/decision-tree-classifier-one-hot-encoder/scikitlearn/default/1/ohe_sex.pkl','rb'))
ohe_embarked = pickle.load(open('/kaggle/input/models/sanjanbm/decision-tree-classifier-one-hot-encoder/scikitlearn/default/1/ohe_embarked.pkl','rb'))
clf = pickle.load(open('/kaggle/input/models/sanjanbm/decision-tree-classifier-one-hot-encoder/scikitlearn/default/1/clf.pkl','rb'))
```

```python
test_input = np.array([2, 'male', 31.0, 0, 0, 10.5, 'S'],dtype=object).reshape(1,7)
```

```python
test_input
```

```python
test_input_sex = ohe_sex.transform(test_input[:,1].reshape(1,1))
```

```python
test_input_sex
```

```python
test_input_embarked = ohe_embarked.transform(test_input[:,-1].reshape(1,1))
```

```python
test_input_embarked
```

```python
test_input_age = test_input[:,2].reshape(1,1)
```

```python
test_input_transformed = np.concatenate((test_input[:,[0,3,4]],test_input_age,test_input_sex,test_input_embarked),axis=1)
```
